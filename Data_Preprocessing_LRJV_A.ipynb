```python
rows = signatures.iloc[1:].values.tolist()
flat = signatures.iloc[1:].values.flatten().tolist()
columns_as_lists = {col: signatures[col].iloc[1:].tolist() for col in signatures.columns}

print(columns_as_lists)
```
